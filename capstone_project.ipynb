```python
import pandas as pd

df = pd.read_csv("issue_tickets_dataset.csv")

print(df.head())
```

```python
cleaned_issues = []

for issue in df["issue_description"]:
    issue = issue.lower()
    issue = issue.replace(".", "")
    issue = issue.replace(",", "")
    issue = issue.replace("!", "")
    issue = issue.replace("?", "")
    issue = issue.replace("-", " ")
    issue = issue.replace("pls", "please")
    issue = " ".join(issue.split())
    issue = issue.strip()

    cleaned_issues.append(issue)


def count_tickets_with_word(word):
    count = 0

    for issue in cleaned_issues:
        if word.lower() in issue.lower():
            count += 1

    return count


print("Tickets containing 'poor':", count_tickets_with_word("poor"))
print("Tickets containing 'good':", count_tickets_with_word("good"))
print("Tickets containing 'slow':", count_tickets_with_word("slow"))
print("Tickets containing 'excellent':", count_tickets_with_word("excellent"))
```

```python
from pprint import pprint


# 1. Display Final Cleaned ticket_data

cleaned_issues = []

for issue in ticket_data["issue"]:

    # Convert to lowercase
    issue = issue.lower()

    # Remove punctuation
    issue = issue.replace(".", "")
    issue = issue.replace(",", "")
    issue = issue.replace("!", "")
    issue = issue.replace("?", "")
    issue = issue.replace("-", " ")

    # Replace slang / shorthand
    issue = issue.replace("ok", "okay")

    # Remove multiple spaces
    issue = " ".join(issue.split())

    # Remove leading/trailing spaces
    issue = issue.strip()

    cleaned_issues.append(issue)


final_ticket_data = {
    "Ticket_No": ticket_data["Ticket_No"],
    "customer_name": ticket_data["customer_name"],
    "issue": cleaned_issues,
    "status": ticket_data["status"]
}

print("1. Final Cleaned ticket_data:")
pprint(final_ticket_data)



# 2. Priority Analysis


high_count = 0
medium_count = 0
low_count = 0

for status in ticket_data["status"]:

    if status.lower() == "high":
        high_count += 1

    elif status.lower() == "medium":
        medium_count += 1

    elif status.lower() == "low":
        low_count += 1


print("\n2. Priority Analysis")
print("Number of High-priority tickets:", high_count)
print("Number of Medium-priority tickets:", medium_count)
print("Number of Low-priority tickets:", low_count)




word_counts = []

for issue in cleaned_issues:
    word_counts.append(len(issue.split()))

max_word_count = max(word_counts)
max_index = word_counts.index(max_word_count)

print("\n3. Ticket With the Longest Issue Description")
print("Ticket number:", ticket_data["Ticket_No"][max_index])
print("Customer name:", ticket_data["customer_name"][max_index])
print("Cleaned issue text:", cleaned_issues[max_index])
print("Word count:", max_word_count)




unique_words = set()

for issue in cleaned_issues:

    words = issue.split()

    for word in words:
        unique_words.add(word)


sorted_words = sorted(unique_words)

print("\n4. Unique Words Used")
print("Count of unique words:", len(sorted_words))

print("Sorted word list:")

for word in sorted_words:
    print(word)
```
